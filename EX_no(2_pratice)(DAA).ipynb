```python
def compute_lps(pattern):
    lps = [0] * len(pattern)
    length = 0
    i = 1
    while i < len(pattern):
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        else:
            if length != 0:
                length = lps[length - 1]
            else:
                lps[i] = 0
                i += 1
    return lps

def kmp_search(text, pattern):
    lps = compute_lps(pattern)
    print(f"LPS Array: {lps}")

    i = 0  # index for text
    j = 0  # index for pattern
    matches = []

    while i < len(text):
        if pattern[j] == text[i]:
            i += 1
            j += 1

        if j == len(pattern):
            matches.append(i - j)
            j = lps[j - 1]
        elif i < len(text) and pattern[j] != text[i]:
            if j != 0:
                j = lps[j - 1]
            else:
                i += 1

    return matches

print("Found matches at indices:", kmp_search("AABCAABXAAAZ", "AABXAA"))
```

```python
def naive_2d_match(text_matrix, pattern_matrix):
    n_rows, n_cols = len(text_matrix), len(text_matrix[0])
    m_rows, m_cols = len(pattern_matrix), len(pattern_matrix[0])

    comparisons = 0
    matches = []

    # Calculate how many starting coordinates are possible
    # For a 10x10 text and 3x3 pattern, these range from 0 to 7 (inclusive)
    for i in range(n_rows - m_rows + 1):
        for j in range(n_cols - m_cols + 1):
            match = True

            # Check the M x M window
            for x in range(m_rows):
                for y in range(m_cols):
                    comparisons += 1  # Increment before checking

                    if text_matrix[i + x][j + y] != pattern_matrix[x][y]:
                        match = False
                        break  # Break inner loop on first mismatch

                if not match:
                    break  # Break outer loop on first mismatch

            if match:
                matches.append((i, j))

    return matches, comparisons

# Example setup
import random

# Generate a random 10x10 matrix of 0s and 1s
T = [[random.randint(0, 1) for _ in range(10)] for _ in range(10)]

# Extract a 3x3 slice to ensure at least one match exists
P = [row[2:5] for row in T[4:7]]

matches, total_comparisons = naive_2d_match(T, P)

print(f"Pattern found at top-left coordinates: {matches}")
print(f"Total element comparisons required: {total_comparisons}")
```
