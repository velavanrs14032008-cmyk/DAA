```python
def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    probes = 0
    while low <= high:
        probes += 1
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid, probes
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, probes


def interpolation_search(arr, target):
    low, high = 0, len(arr) - 1
    probes = 0

    while low <= high and arr[low] <= target <= arr[high]:
        probes += 1
        if arr[high] == arr[low]:
            if arr[low] == target:
                return low, probes
            break

        # Calculate probe index using uniform distribution assumption
        pos = low + int(
            ((target - arr[low]) / (arr[high] - arr[low])) * (high - low)
        )

        if arr[pos] == target:
            return pos, probes
        if arr[pos] < target:
            low = pos + 1
        else:
            high = pos - 1

    return -1, probes


# Roll numbers from 1 to 10,000
roll_numbers = list(range(1, 10001))

# Compare probe counts for sample targets
targets = [1, 2500, 5000, 7843, 9999]
print(f"{'Target':<10}{'Binary Search Probes':<22}{'Interpolation Search Probes'}")
print("-" * 55)
for t in targets:
    _, bs_p = binary_search(roll_numbers, t)
    _, is_p = interpolation_search(roll_numbers, t)
    print(f"{t:<10}{bs_p:<22}{is_p}")
```

```python
import numpy as np


def interpolation_search_float(arr, target):
    low, high = 0, len(arr) - 1
    probes = 0

    while low <= high and arr[low] <= target <= arr[high]:
        probes += 1
        if arr[high] == arr[low]:
            if arr[low] == target:
                return low, probes
            break

        # Float division for position fraction
        fraction = (target - arr[low]) / (arr[high] - arr[low])
        pos = low + int(fraction * (high - low))

        # Boundary safety check
        pos = max(low, min(pos, high))

        if arr[pos] == target:
            return pos, probes
        elif arr[pos] < target:
            low = pos + 1
        else:
            high = pos - 1

    return -1, probes


# Benchmark across different dataset sizes
dataset_sizes = [10000, 50000, 100000]

for size in dataset_sizes:
    # Uniform float distribution between 0.0 and 1000.0
    data = np.sort(np.random.uniform(0.0, 1000.0, size))

    # Test over 1,000 random targets present in the array
    sample_targets = np.random.choice(data, size=1000, replace=False)
    total_probes = sum(
        interpolation_search_float(data, t)[1] for t in sample_targets
    )
    avg_probes = total_probes / 1000

    print(f"Dataset Size N = {size:6d} | Average Probes: {avg_probes:.3f}")
```
