```python
import heapq


def dijkstra_fastest_route(graph, start, target):
    dist = {node: float("inf") for node in graph}
    parent = {node: None for node in graph}

    dist[start] = 0
    pq = [(0, start)]  # (travel_time, node)

    while pq:
        current_time, u = heapq.heappop(pq)

        if current_time > dist[u]:
            continue

        if u == target:
            break

        for v, time in graph[u]:
            if dist[u] + time < dist[v]:
                dist[v] = dist[u] + time
                parent[v] = u
                heapq.heappush(pq, (dist[v], v))

    # Path reconstruction
    path = []
    curr = target
    while curr is not None:
        path.append(curr)
        curr = parent[curr]
    path.reverse()

    return dist[target], path


# 5-node city network
road_network = {
    0: [(1, 4), (2, 2)],
    1: [(2, 1), (3, 5)],
    2: [(3, 8), (4, 10)],
    3: [(4, 2)],
    4: [],
}

travel_time, route = dijkstra_fastest_route(road_network, start=0, target=4)
print(f"Fastest Travel Time: {travel_time} minutes")
print(f"Route: {' -> '.join(map(str, route))}")
```

```python
import heapq


def dijkstra_min_hops(graph, num_nodes, start, target):
    # Track both distance and hop count for every node
    dist = {i: float("inf") for i in range(num_nodes)}
    hops = {i: float("inf") for i in range(num_nodes)}
    parent = {i: None for i in range(num_nodes)}

    dist[start] = 0
    hops[start] = 0

    # Tuple ordering: (total_weight, hop_count, node)
    pq = [(0, 0, start)]

    while pq:
        d, h, u = heapq.heappop(pq)

        # Skip outdated entries
        if d > dist[u] or (d == dist[u] and h > hops[u]):
            continue

        for v, weight in graph[u]:
            new_dist = dist[u] + weight
            new_hops = hops[u] + 1

            # Relax edge if path is shorter OR equal weight with fewer hops
            if new_dist < dist[v] or (
                new_dist == dist[v] and new_hops < hops[v]
            ):
                dist[v] = new_dist
                hops[v] = new_hops
                parent[v] = u
                heapq.heappush(pq, (new_dist, new_hops, v))

    # Reconstruct path
    path = []
    curr = target
    while curr is not None:
        path.append(curr)
        curr = parent[curr]
    path.reverse()

    return dist[target], hops[target], path


# 6-node network graph
graph_6nodes = {
    0: [(1, 2), (3, 3), (4, 1)],
    1: [(2, 2)],
    2: [(5, 2)],  # 0 -> 1 -> 2 -> 5 = weight 6, 3 hops
    3: [(5, 3)],  # 0 -> 3 -> 5      = weight 6, 2 hops
    4: [(5, 7)],  # 0 -> 4 -> 5      = weight 8, 2 hops
    5: [],
}

weight, hop_count, optimal_path = dijkstra_min_hops(
    graph_6nodes, num_nodes=6, start=0, target=5
)

print(f"Total Weight: {weight}")
print(f"Total Hops:   {hop_count}")
print(f"Optimal Path: {' -> '.join(map(str, optimal_path))}")
```
