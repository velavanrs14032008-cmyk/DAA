```python
def find_top_two_min_max(arr, low, high):
    """Returns a tuple: (min1, min2, max1, max2)

    where min1 <= min2 and max2 <= max1.
    """
    # Base Case 1: Single element
    if low == high:
        return (arr[low], float("inf"), arr[low], float("-inf"))

    # Base Case 2: Two elements (1 comparison)
    if high == low + 1:
        if arr[low] < arr[high]:
            return (arr[low], arr[high], arr[high], arr[low])
        else:
            return (arr[high], arr[low], arr[low], arr[high])

    # Divide
    mid = (low + high) // 2
    left_min1, left_min2, left_max1, left_max2 = find_top_two_min_max(
        arr, low, mid
    )
    right_min1, right_min2, right_max1, right_max2 = find_top_two_min_max(
        arr, mid + 1, high
    )

    # Combine Min and 2nd Min (2 comparisons)
    if left_min1 < right_min1:
        g_min1 = left_min1
        g_min2 = min(left_min2, right_min1)
    else:
        g_min1 = right_min1
        g_min2 = min(left_min1, right_min2)

    # Combine Max and 2nd Max (2 comparisons)
    if left_max1 > right_max1:
        g_max1 = left_max1
        g_max2 = max(left_max2, right_max1)
    else:
        g_max1 = right_max1
        g_max2 = max(left_max1, right_max2)

    return (g_min1, g_min2, g_max1, g_max2)


# Example usage
data = [12, 3, 45, 7, 19, 89, 2, 54]
min1, min2, max1, max2 = find_top_two_min_max(data, 0, len(data) - 1)
print(f"1st Min: {min1}, 2nd Min: {min2}")
print(f"1st Max: {max1}, 2nd Max: {max2}")
```

```python
import random


def partition(arr, low, high):
    pivot = arr[high]
    i = low
    for j in range(low, high):
        if arr[j] <= pivot:
            arr[i], arr[j] = arr[j], arr[i]
            i += 1
    arr[i], arr[high] = arr[high], arr[i]
    return i


def quickselect(arr, low, high, k):
    """Finds the 1-based k-th smallest element in arr[low..high]."""
    if low <= high:
        pivot_index = partition(arr, low, high)

        # 1-based rank of the pivot
        rank = pivot_index + 1

        if rank == k:
            return arr[pivot_index]
        elif rank > k:
            return quickselect(arr, low, pivot_index - 1, k)
        else:
            return quickselect(arr, pivot_index + 1, high, k)

    return -1


# Generate test array of 1,000 random integers
random.seed(42)
n = 1000
random_array = [random.randint(1, 100000) for _ in range(n)]

# Sort a reference copy for verification
sorted_reference = sorted(random_array)

# Benchmark positions
test_k_values = {
    "k = 1 (Minimum)": 1,
    f"k = {n//2} (Median)": n // 2,
    f"k = {n} (Maximum)": n,
}

for label, k in test_k_values.items():
    # Pass a fresh copy since QuickSelect partitions in-place
    arr_copy = list(random_array)
    result = quickselect(arr_copy, 0, n - 1, k)
    expected = sorted_reference[k - 1]

    print(f"{label:<20} | Found: {result:<6} | Expected: {expected:<6} | Correct: {result == expected}")
```
