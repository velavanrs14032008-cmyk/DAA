```python
class DisjointSet:

    def __init__(self, vertices):
        self.parent = {v: v for v in vertices}

    def find(self, item):
        if self.parent[item] == item:
            return item
        self.parent[item] = self.find(self.parent[item])  # Path compression
        return self.parent[item]

    def union(self, set1, set2):
        root1 = self.find(set1)
        root2 = self.find(set2)
        if root1 != root2:
            self.parent[root1] = root2
            return True
        return False


def kruskal(vertices, edges):
    # Sort edges by weight
    sorted_edges = sorted(edges, key=lambda edge: edge[2])
    dsu = DisjointSet(vertices)

    mst = []
    total_cost = 0

    for u, v, weight in sorted_edges:
        if dsu.union(u, v):
            mst.append((u, v, weight))
            total_cost += weight

    return mst, total_cost


# Graph definition
vertices = ["A", "B", "C", "D", "E", "F"]
edges = [
    ("A", "B", 4),
    ("A", "C", 2),
    ("B", "C", 5),
    ("B", "D", 10),
    ("C", "E", 3),
    ("D", "E", 7),
    ("D", "F", 8),
    ("E", "F", 6),
]

mst_edges, min_cost = kruskal(vertices, edges)

print("MST Edges:", mst_edges)
print("Total Cost:", min_cost)
```

```python
import heapq
import time
import numpy as np


def prim_matrix(matrix):
    V = len(matrix)
    key = [float("inf")] * V
    parent = [-1] * V
    in_mst = [False] * V

    key[0] = 0  # Start at node 0

    for _ in range(V):
        # Find minimum key vertex from vertices not yet in MST: O(V)
        min_val = float("inf")
        u = -1
        for v in range(V):
            if not in_mst[v] and key[v] < min_val:
                min_val = key[v]
                u = v

        if u == -1:
            break

        in_mst[u] = True

        # Update key values of adjacent vertices: O(V)
        for v in range(V):
            if matrix[u][v] > 0 and not in_mst[v] and matrix[u][v] < key[v]:
                key[v] = matrix[u][v]
                parent[v] = u

    return sum(key), parent


def prim_heap(adj_list, V):
    key = [float("inf")] * V
    parent = [-1] * V
    in_mst = [False] * V

    key[0] = 0
    pq = [(0, 0)]  # (key, vertex)
    total_cost = 0

    while pq:
        w, u = heapq.heappop(pq)

        if in_mst[u]:
            continue

        in_mst[u] = True
        total_cost += w

        for v, weight in adj_list[u]:
            if not in_mst[v] and weight < key[v]:
                key[v] = weight
                parent[v] = u
                heapq.heappush(pq, (weight, v))

    return total_cost, parent
```
